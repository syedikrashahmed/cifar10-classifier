# intro to tensors and tensor operations in pytorch


```python
import torch
import numpy as np
```

initializing a tensor

```python
data = [[1, 2],[3, 4]]
x_data = torch.tensor(data)
x_data
```

```python
np_array = np.array(data)
x_np = torch.from_numpy(np_array)
x_np
```

```python
x_ones = torch.ones_like(x_data) # retains the properties of x_data
print(f"Ones Tensor: \n {x_ones} \n")

x_rand = torch.rand_like(x_data, dtype=torch.float) # overrides the datatype of x_data
print(f"Random Tensor: \n {x_rand} \n")
```

```python
shape = (1,3)
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)

print(f"Random Tensor: \n {rand_tensor} \n")
print(f"Ones Tensor: \n {ones_tensor} \n")
print(f"Zeros Tensor: \n {zeros_tensor}")
```

properties of a tensor

```python
tensor = torch.rand(3,4)

print(f"Shape of tensor: {tensor.shape}")
print(f"Datatype of tensor: {tensor.dtype}")
print(f"Device tensor is stored on: {tensor.device}")
```

switching to an accelerator instead of cpu

```python
# We move our tensor to the current accelerator if available
if torch.accelerator.is_available():
    tensor = tensor.to(torch.accelerator.current_accelerator())
    print(tensor.device)
```

tensor operations

```python
tensor = torch.ones(4, 4)
print(tensor)
print(f"First row: {tensor[0]}")
print(f"First column: {tensor[:, 0]}")
print(f"Last column: {tensor[..., -1]}")
tensor[:,1] = 0
print(tensor)
```

```python
t1 = torch.cat([tensor, tensor, tensor], dim=1)
print(t1)
```

arithmetic operations

```python
# This computes the matrix multiplication between two tensors. y1, y2, y3 will have the same value
# ``tensor.T`` returns the transpose of a tensor
y1 = tensor @ tensor.T
y2 = tensor.matmul(tensor.T)

y3 = torch.rand_like(y1)
torch.matmul(tensor, tensor.T, out=y3)
print(f"y1: {y1}")
print(f"y2: {y2}")
print(f"y3: {y3}")
print()

# This computes the element-wise product. z1, z2, z3 will have the same value
z1 = tensor * tensor
z2 = tensor.mul(tensor)
z3 = torch.rand_like(tensor)
torch.mul(tensor, tensor, out=z3)
print(f"z1: {z1}")
print(f"z2: {z2}")
print(f"z3: {z3}")
```

```python
agg = tensor.sum()
print(agg.item(), type(agg.item()))
```

```python
print(f"{tensor} \n")
tensor.add_(5)
print(tensor)
```

Bridge with NumPy:
Tensors on the CPU and NumPy arrays can share their underlying memory locations, and changing one will change the other.

```python
t = torch.ones(5)
print(f"t: {t}")
n = t.numpy()
print(f"n: {n}")
```

```python
t.add_(1)
print(f"t: {t}")
print(f"n: {n}")
```

```python
n = np.ones(5)
t = torch.from_numpy(n)
print(f"n: {n}")
print(f"t: {t}")
```

```python
np.add(n, 1, out=n)
print(f"t: {t}")
print(f"n: {n}")
```

create vectors <br> create matrices <br> perform matrix multiplication <br> reshape tensors <br> transpose them <br> compute mean <br> compute sum <br> compute max <br> generate random tensors

```python
# Scalar (0D tensor)
scalar = torch.tensor(7)
print("Scalar:")
print(scalar)
print("Shape:", scalar.shape)

# Vector (1D tensor)
vector = torch.tensor([1, 2, 3, 4, 5])
print("\nVector:")
print(vector)
print("Shape:", vector.shape)

# Matrix (2D tensor)
matrix = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

print("\nMatrix:")
print(matrix)
print("Shape:", matrix.shape)
```

```python
A = torch.tensor([
    [1, 2],
    [3, 4]
])

B = torch.tensor([
    [5, 6],
    [7, 8]
])

print("Matrix A:\n", A)
print("\nMatrix B:\n", B)

# Matrix multiplication
C = torch.matmul(A, B)

print("A @ B =")
print(C)
```

```python
#can also do this
C = A @ B
print(C)
```

```python
tensor = torch.arange(12)

print("Original:")
print(tensor)
print("Shape:", tensor.shape)

reshaped = tensor.reshape(3, 4)

print("\nReshaped:")
print(reshaped)
print("Shape:", reshaped.shape)
```

```python
matrix = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

print("Original:")
print(matrix)

print("\nTranspose:")
print(matrix.T)
```

```python
values = torch.tensor([10., 20., 30., 40., 50.])

print("Values:", values)
print("Mean:", values.mean())
print("Sum:", values.sum())
print("Maximum:", values.max())
print("Index of max:", values.argmax())
```

```python
random_tensor = torch.rand(3, 4)

print(random_tensor)
```

```python
random_ints = torch.randint(
    low=0,
    high=10,
    size=(3, 4)
)

print(random_ints)
```

```python
#rand from a normal distribution
random_normal = torch.randn(3, 4)

print(random_normal)
```

```python
print("Zeros:")
print(torch.zeros(2, 3))

print("\nOnes:")
print(torch.ones(2, 3))

print("\nIdentity Matrix:")
print(torch.eye(4))

print("\nSequence:")
print(torch.arange(0, 10))

print("\nEvenly spaced values:")
print(torch.linspace(0, 1, 5))
```
